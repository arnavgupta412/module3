### Writing your OpenRouter code for the OpenAI model

Use OpenRouter and the `OPENAI_MODEL` model variable.


````python
# --- Student Function: OpenAI through OpenRouter ---
from openai import OpenAI

def call_openai(prompt: str) -> str:
    """Return the plain-text response from the selected OpenAI model."""
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)
    response = client.chat.completions.create(
        model=OPENAI_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
        max_tokens=250,
    )
    return response.choices[0].message.content or ""
````

````python
import json
import re
import time
import pandas as pd

CREATIVE_PROMPT = "Write a 4-line poem about learning by testing code."
ANALYTICAL_PROMPT = (
    "Summarize in one sentence (<= 40 words): "
    "'Using multiple LLM providers helps compare creativity, formatting reliability, and speed, "
    "but adds integration overhead and testing complexity.'"
)
INSTRUCTION_PROMPT = (
    "Return ONLY valid JSON (no backticks). The JSON must be an object with key 'todos' "
    "which is a list of EXACTLY 2 items. Each item must be an object with keys 'task' (string) "
    "and 'done' (boolean)."
)

MODELS = {
    "OpenAI GPT-4o-mini": call_openai,
    "Gemini 2.5 Flash": call_gemini,
    "Claude Sonnet 4.6": call_claude,
}

TASKS = {
    "Creative Writing": CREATIVE_PROMPT,
    "Analytical Reasoning": ANALYTICAL_PROMPT,
    "Instruction Following": INSTRUCTION_PROMPT,
}

def valid_creative(text):
    return len([x for x in text.splitlines() if x.strip()]) >= 4

def valid_analytical(text):
    words = re.findall(r"\S+", text.strip())
    sentences = [x for x in re.split(r"(?<=[.!?])\s+", text.strip()) if x]
    return bool(text.strip()) and len(words) <= 40 and len(sentences) == 1 and bool(re.search(r"[.!?]\s*$", text.strip()))

def valid_json(text):
    try:
        obj = json.loads(text)
        return (
            set(obj.keys()) == {"todos"} and
            isinstance(obj["todos"], list) and
            len(obj["todos"]) == 2 and
            all(
                isinstance(x, dict) and set(x.keys()) == {"task", "done"} and
                isinstance(x["task"], str) and isinstance(x["done"], bool)
                for x in obj["todos"]
            )
        )
    except Exception:
        return False

VALIDATORS = {
    "Creative Writing": valid_creative,
    "Analytical Reasoning": valid_analytical,
    "Instruction Following": valid_json,
}

results = []
for task, prompt in TASKS.items():
    print("\n" + "=" * 70)
    print(task)
    print("Prompt:", prompt)
    for model_name, fn in MODELS.items():
        t0 = time.perf_counter()
        output = fn(prompt)
        latency = time.perf_counter() - t0
        passed = VALIDATORS[task](output)
        results.append({
            "Task": task,
            "Model": model_name,
            "Latency (s)": round(latency, 3),
            "Format": "PASS" if passed else "FAIL",
            "Output": output,
        })
        print(f"\n{model_name} | {latency:.3f}s | {'PASS' if passed else 'FAIL'}")
        print(output)

results_df = pd.DataFrame(results)
display(results_df)
````

````python
# Task-level comparison summary
summary = results_df.copy()
summary["Pass"] = summary["Format"] == "PASS"

display(summary.pivot(index="Task", columns="Model", values=["Latency (s)", "Format"]))

avg_latency = summary.groupby("Model")["Latency (s)"].mean().sort_values()
format_rate = (summary.groupby("Model")["Pass"].mean() * 100).round(1).sort_values(ascending=False)

print("Average latency (lower is faster):")
display(avg_latency.to_frame("Average latency (s)"))

print("Format compliance: " )
display(format_rate.to_frame("Pass rate (%)"))
````
