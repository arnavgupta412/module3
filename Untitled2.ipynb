```python
# ============================================
# CELL 1 — INSTALL, IMPORTS & API SETUP
# ============================================

!pip install -q pypdf openai pinecone

import os
import re
import time
import numpy as np

from getpass import getpass
from pypdf import PdfReader
from openai import OpenAI
from pinecone import Pinecone, ServerlessSpec

# Enter your keys when prompted
OPENROUTER_API_KEY = getpass("Enter your OpenRouter API key: ")
PINECONE_API_KEY = getpass("Enter your Pinecone API key: ")

# Environment variables
os.environ["OPENAI_API_KEY"] = OPENROUTER_API_KEY
os.environ["PINECONE_API_KEY"] = PINECONE_API_KEY

# OpenRouter
client = OpenAI(
    api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)

# Pinecone
pc = Pinecone(api_key=PINECONE_API_KEY)

INDEX_NAME = "pdf-mini-app"

print("Setup complete!")
```

```python
# ============================================
# CELL 2 — REQUIRED RAG FUNCTIONS
# ============================================

def parse_pdf_to_chunks(pdf_path, chunk_size=800, overlap=120):
    """Extract PDF text and create deterministic overlapping chunks."""

    if overlap >= chunk_size:
        raise ValueError("overlap must be smaller than chunk_size")

    reader = PdfReader(pdf_path)
    texts = []

    for page in reader.pages:
        text = page.extract_text() or ""
        text = re.sub(r"\s+", " ", text).strip()

        if text:
            texts.append(text)

    full_text = " ".join(texts)

    if not full_text:
        return []

    chunks = []
    start = 0

    while start < len(full_text):
        end = min(start + chunk_size, len(full_text))
        chunk = full_text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        if end >= len(full_text):
            break

        start = end - overlap

    return chunks


def embed_and_index(
    chunks,
    use_pinecone=True,
    model="text-embedding-3-small"
):
    """Create embeddings with OpenRouter and store them in Pinecone."""

    if not chunks:
        raise ValueError("No chunks were provided.")

    # OpenRouter model ID
    embedding_model = model

    if model == "text-embedding-3-small":
        embedding_model = "openai/text-embedding-3-small"

    # Generate embeddings
    response = client.embeddings.create(
        model=embedding_model,
        input=chunks
    )

    embeddings = np.array(
        [item.embedding for item in response.data],
        dtype=np.float32
    )

    dimension = embeddings.shape[1]

    print("Embedding dimension:", dimension)

    # Check existing Pinecone indexes
    existing = pc.list_indexes()

    try:
        existing_names = existing.names()
    except AttributeError:
        existing_names = [x["name"] for x in existing]

    # Create index if necessary
    if INDEX_NAME not in existing_names:

        print("Creating Pinecone index...")

        pc.create_index(
            name=INDEX_NAME,
            dimension=dimension,
            metric="cosine",
            spec=ServerlessSpec(
                cloud="aws",
                region="us-east-1"
            )
        )

        while True:
            info = pc.describe_index(INDEX_NAME)

            try:
                ready = info.status["ready"]
            except:
                ready = info.status.ready

            if ready:
                break

            time.sleep(2)

    index = pc.Index(INDEX_NAME)

    # Stable IDs + metadata
    vectors = []

    for i, (chunk, embedding) in enumerate(
        zip(chunks, embeddings)
    ):
        vectors.append({
            "id": f"doc-{i}",
            "values": embedding.tolist(),
            "metadata": {
                "chunk": chunk,
                "order": i
            }
        })

    index.upsert(vectors=vectors)

    return {
        "chunks": chunks,
        "index_backend": "pinecone",
        "index_ref": INDEX_NAME,
        "embed_model": embedding_model,
        "dim": dimension,
        "id_to_chunk": {
            f"doc-{i}": chunk
            for i, chunk in enumerate(chunks)
        }
    }


def answer_query(
    question,
    index_obj,
    top_k=3,
    gpt_model="gpt-4o-mini"
):
    """Retrieve PDF chunks and generate a grounded answer."""

    # Embed question using same embedding model
    response = client.embeddings.create(
        model=index_obj["embed_model"],
        input=[question]
    )

    query_embedding = response.data[0].embedding

    # Pinecone search
    index = pc.Index(index_obj["index_ref"])

    results = index.query(
        vector=query_embedding,
        top_k=top_k,
        include_metadata=True
    )

    # Build context
    context_chunks = []

    for match in results["matches"]:
        metadata = match.get("metadata", {})
        chunk = metadata.get("chunk", "")

        if chunk:
            context_chunks.append(chunk)

    context = "\n\n---\n\n".join(context_chunks)

    # Required context-only prompt
    system_prompt = """
You are a helpful assistant. Use ONLY the provided context to answer.

If not answerable from context, respond:
"I don't know based on the document."
""".strip()

    user_prompt = f"""
Question:
{question}

Context:
{context}
""".strip()

    # OpenRouter model
    actual_model = gpt_model

    if gpt_model == "gpt-4o-mini":
        actual_model = "openai/gpt-4o-mini"

    response = client.chat.completions.create(
        model=actual_model,
        messages=[
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": user_prompt
            }
        ],
        temperature=0
    )

    return response.choices[0].message.content.strip()


def vanilla_answer(question):
    """Answer without PDF retrieval."""

    response = client.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {
                "role": "system",
                "content": "You are a helpful assistant."
            },
            {
                "role": "user",
                "content": question
            }
        ],
        temperature=0
    )

    return response.choices[0].message.content.strip()


print("All RAG functions are ready!")
```

```python
# ============================================
# CELL 3 — UPLOAD DOCX/PDF + RAG + COMPARISON
# ============================================

from google.colab import files
import os
import subprocess

# --------------------------------------------
# Upload your document
# --------------------------------------------

uploaded = files.upload()

filename = list(uploaded.keys())[0]
input_path = "/content/" + filename

print("\nUploaded:", filename)


# --------------------------------------------
# Convert DOCX to PDF if necessary
# --------------------------------------------

if filename.lower().endswith(".docx"):

    print("\nDOCX detected. Converting to PDF...")

    !apt-get -qq update
    !apt-get -qq install -y libreoffice

    subprocess.run([
        "libreoffice",
        "--headless",
        "--convert-to",
        "pdf",
        "--outdir",
        "/content",
        input_path
    ], check=True)

    PDF_PATH = (
        "/content/" +
        os.path.splitext(filename)[0] +
        ".pdf"
    )

elif filename.lower().endswith(".pdf"):

    PDF_PATH = input_path

else:
    raise ValueError("Please upload a PDF or DOCX file.")

print("Using PDF:", PDF_PATH)


# --------------------------------------------
# Parse PDF
# --------------------------------------------

chunks = parse_pdf_to_chunks(
    PDF_PATH,
    chunk_size=800,
    overlap=120
)

print("\nNumber of chunks:", len(chunks))

if not chunks:
    raise ValueError("No text could be extracted from the PDF.")

print("\nFirst chunk preview:")
print(chunks[0][:500])


# --------------------------------------------
# Create Pinecone index
# --------------------------------------------

print("\nCreating embeddings and indexing...")

index_obj = embed_and_index(
    chunks,
    use_pinecone=True,
    model="text-embedding-3-small"
)

print("\nPinecone indexing complete!")
print("Index:", index_obj["index_ref"])
print("Dimension:", index_obj["dim"])


# ============================================
# 3 DOMAIN-SPECIFIC QUESTIONS
# ============================================

questions = [
    "What is the main objective of The Lost Crystal of Eldoria?",

    "What are the three main branching paths in the game and what is the outcome of each path?",

    "What were the main strengths and weaknesses identified during the three playtests?"
]


# ============================================
# VANILLA LLM VS RAG
# ============================================

for i, question in enumerate(questions, 1):

    print("\n" + "=" * 80)
    print(f"QUESTION {i}")
    print("=" * 80)

    print(question)

    # Vanilla LLM
    print("\n--- VANILLA LLM ANSWER ---")

    vanilla = vanilla_answer(question)

    print(vanilla)

    # RAG
    print("\n--- RAG ANSWER ---")

    rag = answer_query(
        question,
        index_obj,
        top_k=3,
        gpt_model="gpt-4o-mini"
    )

    print(rag)


print("\n" + "=" * 80)
print("RAG PLAYTEST COMPLETED")
print("=" * 80)
```
